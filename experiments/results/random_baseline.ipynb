## Reproduce the results

Run the cell below to regenerate the summary and distributions. Run `python -m experiments.random_baseline` from the repository root to regenerate the slide figure.

```python
from collections import Counter
from pathlib import Path
from statistics import mean
import sys

cwd = Path.cwd()
repo = next(path for path in (cwd, *cwd.parents) if (path / "agents" / "random_agent.py").is_file())
sys.path.insert(0, str(repo))

from agents.random_agent import RandomAgent
from game import play_games

results = play_games(RandomAgent(seed=0), seeds=list(range(100)))
total_moves = sum(game.moves for game in results)
mean_move_ms = 1000 * sum(game.total_time for game in results) / total_moves

print(f"Games: {len(results)}")
print(f"Mean final score: {mean(game.score for game in results):,.2f}")
print(f"Highest tile: {max(game.max_tile for game in results)}")
print(f"Games reaching 2048: {sum(game.reached_2048 for game in results)}")
print(f"Mean decision time: {mean_move_ms:.3f} ms per move")

score_bins = Counter(game.score // 400 for game in results)
for index, count in sorted(score_bins.items()):
    print(f"Score {index * 400:,}–{index * 400 + 399:,}: {count} games")

for tile, count in sorted(Counter(game.max_tile for game in results).items()):
    print(f"Highest tile {tile}: {count} games")
```
